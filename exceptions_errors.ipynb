```python
import math
try:
    num1 = 10
    num2 = int(input("Enter a number: "))
    res = num1 / num2
except Exception as e:
    print(e)
    if isinstance(e, ZeroDivisionError):
        res = math.inf
    elif isinstance(e, ValueError):
        num2 = math.nan
        res = math.nan
else:
    print("---- this block will be executed if you don't have any exceptions... ")
    print(f"num1 = {num1}, num2 = {num2}, res = {res}")
```

```python
def div_nums():
    try:
        num1 = int(input("Enter a number: "))
        num2 = int(input("Enter a number: "))
        res = num1 / num2
    except Exception as e:
        print("---- error ----")
        print(e)
        return  math.nan
    else:
        print("----success---")
        print(f"num1 = {num1}, num2 = {num2}, res = {res}")
        return res
    finally:
        # finally execution precedes return
        print("-----thank you for using this function")
    print("------**************************")
```
