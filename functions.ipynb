```python
def sum_nums2(num1, num2):
    # isdigit()
    if num1.isdigit() and num2.isdigit():
        num1 = int(num1)
        num2 = int(num2)
        print(f"num1: {num1}, num2: {num2}")
        res = num1 + num2
        print(res)


sum_nums2(3,4)
```

```python
def sum_nums2(num1, num2):
    if type(num1) == int and type(num2) == int:
        print(f"num1: {num1}, num2: {num2}")
        res = num1 + num2
        print(res)
    else:
        print(f"num1: {num1}, num2: {num2} are not valid integers")


sum_nums2(3,"4")
```

```python
def sum_nums2 (num1 :int , num2 : int ) -> None:
    """
    this function calculates the sum of two numbers
    :param num1:  is integer
    :param num2:  is integer
    :return:  is integer
    """
    if isinstance(num1, int) and isinstance(num2, int):
        print(f"num1: {num1}, num2: {num2}")
        res = num1 + num2
        print(res)
    else:
        print(f"num1: {num1}, num2: {num2} are not valid integers")


sum_nums2(3,"4")
```
