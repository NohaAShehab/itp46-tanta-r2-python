```python
# map, filter

nums = [3,4,5,6]
new_nums = map(lambda x : x * 2, nums)
print(new_nums) # map_object can be cased to a list
print(list(new_nums))
```

```python
# filter out odd numbers
nums = [3,4,5,6]
odd_nums = filter(lambda x : x % 2 != 0, nums)
print(odd_nums)
print(list(odd_nums))
```
