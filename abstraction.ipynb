```python
"""

    abstract class --> cannot take object from this class
    define base structure =-> should inherited from the child class

    abstract classes must have at least one abstract method
    use keyword abstract  --> abstract class Person{}
"""
"""
    in python to define an abstract class , you need to inherit from
    class ABC (Abstract Base Class) and has at least one abstract method
"""
from abc import ABC, abstractmethod

class Person(ABC):
    @abstractmethod
    def speak(self):pass

p = Person()


```

```python
class Person(ABC):
    @abstractmethod
    def speak(self):pass

class Teacher(Person):
    def speak(self):
        print("---- I am a teacher")

t = Teacher()
t.speak()
print(isinstance(t, Person))
```
