```python
import json
try:
    with open("users.json", "r") as f:
        users = json.load(f)
        print(users, type(users))
except Exception as e:
    users= []
```

```python
def read_all_data():
    try:
        with open("users.json", "r") as f:
            users = json.load(f)
            print(users, type(users))

    except Exception as e:
        users= []
    return  users
```

```python
def save_data(users: list):
     try:
        with open("users.json", "w") as f:
            json.dump(users, f, indent=4)
            print(users, type(users))

     except Exception as e:
         users= []
         return  users
```

```python
user1= {"name": "iti", "id":4,
        "track": "python"}

allusers = read_all_data() # list
allusers.append(user1)
save_data(allusers)
```
