```python
# comma separated key value pair ?
info = {
    "name":"noha,",
    "track": "python",
    "city": "tanta",
    "name":"Noha"
}

print(info )
# before python 3.6 --> dict like sets , un-ordered datatype #
# now dict is ordered datatype
```

```python
# access element using keyword
print(info['name'])
```

```python
# dict is mutable datatype
info['name'] = 'Noha Shehab'
print(info)
```

```python
info['salary']= 1000 # key doesn't exist , add it
print(info)
```

```python
print(len(info))
```

```python
for item in info:
    print(item)
```

```python
for item , val in info:
    print(item ,val)
```

```python
for item in info:
    print(f'{item}: {info[item]}')
```

```python
# get keys of dicts
info_keys = info.keys()
print(info_keys) # could be casted to list
```

```python
info_vals = info.values()
print(info_vals, type(info_vals))
```

```python
info_items = info.items()
for k, v in info_items:
    print(f"{k}: {v}")
```

```python
for i in info_items:
    print(i, type(i))
```
