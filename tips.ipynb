# sets

```python
mys = {"iti", "iti", 323, 3123.32, "iti", "ali", 'django',
       'abc', "iti", 'ali', 'abc'}

print(mys)
```

```python
# 1- remove duplication
# 2- no order
# 3- no index

# set is mutable , you can add values to it.
mys.add("new element")
print(mys)
```

```python
mys.add("python")
print(mys)
```

```python
# save data in sets in hashed
mys = {"iti", "iti", 323, 3123.32, "iti", "ali", 'django',
       'abc', "iti", 'ali', 'abc'}

mys.add([34,32])
```

```python
mys.add((32, 33))
```

```python
print(mys)
```
