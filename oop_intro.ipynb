# Instance variables

```python
# customize object creation ?
class Employee:
    def __init__(self, name, email, salary):
        # name, email  , salary --> instance variables ?
        self.name=name
        self.email = email
        self.salary = salary

emp = Employee("Ahmed", "ahmed@gmail.com", 344)
print(emp)
print(emp.__dict__)
```

```python
"""
    instance variables >>> values of these variables depend on the caller instance
"""

class Employee:
    def __init__(self, name, email, salary):
        # name, email  , salary --> instance variables ?
        self.name=name
        self.email = email
        self.salary = salary

emp = Employee("Ahmed", "ahmed@gmail.com", 344)
emp2 = Employee("test", 'test@gmail.com', 2231)
print(emp.name)
print(emp2.name)
```

# instance methods

```python
# define instance methods ? => functionalities could be applied on the instances / objects

class Employee:
    def __init__(self, name, email, salary):
        # name, email  , salary --> instance variables ?
        self.name=name
        self.email = email
        self.salary = salary

    # define action, activity ?
    # define instance method
    def display_employee(self):
        print(f"name={self.name}, email={self.email}, salary={self.salary}")
```

```python
emp = Employee("ali", "ali@gmail.com", 224)
emp.display_employee()
```

```python
emp2= Employee('new', "new@gmail.com", 1224)
emp2.display_employee()
```

# class method

```python
class Employee:
    no_of_employees = 0
    country = "Egypt"
    def __init__(self, name, email, salary):
        self.name=name
        self.email = email
        self.salary = salary
        Employee.no_of_employees +=1

    # function >> return with no_of_employees
    # represent class behaviour
    @classmethod # this method represent class beh.
    def get_no_of_employees(cls):
        print(cls)
        return Employee.no_of_employees



print(Employee.get_no_of_employees())
```

```python
class Employee:
    no_of_employees = 0
    country = "Egypt"
    def __init__(self, name, email, salary):
        self.name=name
        self.email = email
        self.salary = salary
        print(self.__class__) # class name
        # Employee.no_of_employees +=1
        self.__class__.no_of_employees +=1

    # function >> return with no_of_employees
    # represent class behaviour
    @classmethod # this method represent class beh.
    def get_no_of_employees(cls):
        print(cls)
        return Employee.no_of_employees

print(Employee.get_no_of_employees())
```

# static methods

```python
class Employee:
    no_of_employees = 0
    country = "Egypt"
    def __init__(self, name, email, salary):
        self.name=name
        self.email = email
        self.salary = salary
        self.__class__.no_of_employees +=1

emp =  Employee("new", "new", 10000)

def cal_net_salary(salary):
    return salary * .8

print(cal_net_salary(emp.salary))
```

```python
cal_net_salary(324325353535)
```

```python
class Employee:
    no_of_employees = 0
    country = "Egypt"
    def __init__(self, name, email, salary):
        self.name=name
        self.email = email
        self.salary = salary
        self.__class__.no_of_employees +=1

    # helper method ?  this method, doesn't depend on object or class , it is helper.
    @staticmethod
    def cal_net_salary(salary):
        return salary * .8

emp = Employee("new", "new", 50000)
emp.cal_net_salary(emp.salary)

print(Employee.cal_net_salary(434346))
```

```python
def validate_str(any_str):
    if isinstance(any_str, str) and len(any_str.strip()) > 0:
        return  any_str
    raise ValueError("not valid")
```

```python
class Student:
    def __init__(self, name):
        self.name = validate_str(name)

s = Student("Ahmed")
```

```python
ss = Student(324)
```

```python
class Student:
    def __init__(self, name):
        self.name = self.validate_str(name)

    @staticmethod
    def validate_str(any_str): # this parm is not self or cls
        if isinstance(any_str, str) and len(any_str.strip()) > 0:
            return  any_str
        raise ValueError("not valid")

s = Student("Ahmed")
```

```python
ss = Student(243.32)
```
