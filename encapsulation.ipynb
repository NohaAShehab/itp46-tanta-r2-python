# why we use private members

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        self.__salary = salary  # private


    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, int) or isinstance(salary, float) and salary> 0:
            self.__salary = salary
        else:
            raise ValueError("Salary must be an integer or float and > 0")

emp = Employee("new", "", 10000)
print(emp.get_salary())
emp.set_salary("any value")
```

# Property

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        self.set_salary(salary)


    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, int) or isinstance(salary, float) and salary> 0:
            self.__salary = salary
        else:
            raise ValueError("Salary must be an integer or float and > 0")

emp = Employee("new", "",1000)
print(emp.get_salary()) # read value of private member __salary
print(emp.set_salary(323)) # set new value of private member __salary
```

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        self.set_salary(salary)

    @property
    def salary(self): # read value from object
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, int) or isinstance(salary, float) and salary> 0:
            self.__salary = salary
        else:
            raise ValueError("Salary must be an integer or float and > 0")

emp = Employee("new", "",1000)
print(emp.__dict__)
###
print(emp.salary) # salary --> function ,, property
```

```python
emp.salary = 5000
```

```python
# I need to allow setting salary ???
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        self.salary= salary

    @property
    def salary(self): # read value from object
        return self.__salary

    @salary.setter
    def salary(self, salary):
        if isinstance(salary, int) or isinstance(salary, float) and salary> 0:
            self.__salary = salary
        else:
            raise ValueError("Salary must be an integer or float and > 0")

    @salary.deleter
    def salary(self):
        del self.__salary



emp = Employee("new", "",1000)
del emp.salary
```
