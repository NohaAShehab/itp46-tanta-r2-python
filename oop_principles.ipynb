```python
#calling parent constructor
class Parent:
    def __init__(self, name):
        print("I am a parent object ")
        self.name = name

    def display(self):
        print(f"My name is {self.name}")

class Child(Parent):
    def __init__(self, name, salary):
        super().__init__(name) # call parent constructor.
        self.salary = salary
        print("I am a child object ")

    # modify behaviour of display function?
    def display(self):
        print(f"salary = {self.salary}")

c = Child("Ahmed", 5000)
c.display()
```

```python
#calling parent constructor
class Parent:
    def __init__(self, name):
        print("I am a parent object ")
        self.name = name

    def display(self):
        print(f"My name is {self.name}")

class Child(Parent):
    def __init__(self, name, salary):
        super().__init__(name) # call parent constructor.
        self.salary = salary
        print("I am a child object ")

    # modify behaviour of display function?
    # override
    def display(self):
        super().display() # call parent display
        print(f"salary = {self.salary}")

c = Child("Ahmed", 5000)
c.display()
```
