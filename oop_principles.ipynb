```python
#calling parent constructor
class Parent:
    def __init__(self, name):
        print("I am a parent object ")
        self.name = name

    def display(self):
        print(f"My name is {self.name}")

class Child(Parent):
    def __init__(self, name, salary):
        super().__init__(name) # call parent constructor.
        self.salary = salary
        print("I am a child object ")

    # modify behaviour of display function?
    # override
    def display(self):
        super().display() # call parent display
        print(f"salary = {self.salary}")

c = Child("Ahmed", 5000)
c.display()
```

# Polymorphism
- overloading

```python
class Employee:
    def __init__(self,name, email):
        self.name = name
        self.email = email

    def print_emp(self, msg):
        print(f"{self.name} {self.email}, {msg}")

    def print_emp(self):
        print(f"{self.name} {self.email}")

emp = Employee("ahmed", "ahmed@gmail.com")
emp.print_emp("Hello from emp")
```

```python
from typing import  overload

@overload
def process(var : str) -> None: pass

@overload
def process(var : int) -> None: pass

def process(var):
    if isinstance(var, int):
        return var * 10
    elif isinstance(var, str):
        return var.upper()
    else:
        return None


print(process(1))
print(process("iti"))
print(process([1, 2, 3]))


```

```python
# multidispatch
from multipledispatch import dispatch
@dispatch(int, int)
def sum_num(num1 : int, num2 : int) -> int:
    return num1 + num2
@dispatch(int, int, int)
def sum_num(num1: int, num2: int, num3: int ) -> int:
    return num1 + num2 + num3

sum_num(3,4)
```

```python
from typing import overload
class Employee:
    def __init__(self,name, email):
        self.name = name
        self.email = email

    @overload
    def print_emp(self, msg: int):pass

    @overload
    def print_emp(self, msg: str):pass

    def print_emp(self, msg):
        if isinstance(msg, str):
            print(msg.upper())
        elif isinstance(msg, int):
            print(msg*10)
        else:
            print("Not Applicable")
e = Employee("sd",33)
e.print_emp("Hello")
e.print_emp(10)
```
