read data

```python
try:
    # 1- open file
    file_obj = open("users.txt", 'r')
    print(file_obj)

    # 2- read data
    data = file_obj.read() # read file content into one string
    print(data, type(data))


    # 3- close file
    file_obj.close()
except Exception as e:
    print(e)
```

```python
try:
    # 1- open file
    file_obj = open("users.txt", 'r')
    print(file_obj)

    # 2- read data line by line
    lines = file_obj.readlines() # list
    print(lines, type(lines))


    # 3- close file
    file_obj.close()
except Exception as e:
    print(e)
```

```python
try:
    # 1- open file
    file_obj = open("users.txt", 'r')
    print(file_obj)

    # 2- read data line by line
    lines = file_obj.readlines() # list
    print(lines, type(lines)) # pointer of object reached end of the file
    print("---------------------------------")
    # reset the pointer to position at byte x
    file_obj.seek(0)

    data = file_obj.read()
    print(data, type(data))


    # 3- close file
    file_obj.close()
except Exception as e:
    print(e)
```

Write data to the file ?

```python
""" when you try to open file with w mode, if file doesn't exists ?
python will try to create  if exists , open file for writing starting
from the beginning of the file, remove old content"""
try:
    # 1- open file
    file_obj2 = open("students.txt",'w')
    print(file_obj2)


    # 2- close file
    file_obj2.close()
except Exception as e:
    print(e)
```

```python
try:
    # 1- open file
    file_obj2 = open("students.txt",'w')
    print(file_obj2)
    file_obj2.write("Hello from python\n")
    file_obj2.write("Hello from application")
    # 2- close file
    file_obj2.close()
except Exception as e:
    print(e)
```

# Append

```python
""" when you try to open file with a mode, if file doesn't exists ?
python will try to create  if exists , open file for writing starting
from the end of the file, keep old content"""
try:
    # 1- open file
    file_obj2 = open("students.txt",'a')
    print(file_obj2)
    file_obj2.writelines("Append new data to students \n")

    # 2- close file
    file_obj2.close()
except Exception as e:
    print(e)
```
